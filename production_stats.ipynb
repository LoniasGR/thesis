```python
import sqlite3

conn = sqlite3.connect("production.db")
```

```python
tables = conn.execute(
    "SELECT name FROM sqlite_master WHERE type='table' ORDER BY name"
).fetchall()

for (table_name,) in tables:
    print(table_name)
```

```python
for name in (
    "clients",
    "evaluations",
    "evaluation_user_dialogue_table",
    "user_dialogues",
):
    print(f"\n{name}:")
    for column in conn.execute(f'PRAGMA table_info("{name}")').fetchall():
        print(column)

    print("Sample rows:")
    for row in conn.execute(f'SELECT * FROM "{name}" LIMIT 4').fetchall():
        print(row)
```

```python
for column in conn.execute(
    'SELECT COUNT(*) FROM "clients" WHERE host NOT IN ("localhost", "127.0.0.1", "test.lavdelas.me", "localhost:8123", "mock-dialogues-service")'
).fetchall():
    print(column)
```

```python
# Count the average number of client_ids per evaluation
for row in conn.execute(
    "SELECT MEDIAN(client_count) FROM (SELECT COUNT(client_id) AS client_count FROM evaluations WHERE answer IS NOT NULL GROUP BY client_id)"
).fetchall():
    print(row)
```

```python
for row in conn.execute(
    "SELECT client_id, COUNT(*) FROM evaluations WHERE answer IS NOT NULL GROUP BY client_id"
).fetchall():
    print(row)
```

```python
for row in conn.execute("SELECT * FROM evaluations WHERE client_id = 4").fetchall():
    print(row)
```
